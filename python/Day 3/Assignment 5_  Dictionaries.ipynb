```python
# Questions

a) Find the student(s) who received grade A.

b) Find the student(s) who received grade C.
```

```python
# 12. Student Attendance
```

```python
attendance = {
    "John": True,
    "Alice": False,
    "Bob": True,
    "David": None
}
```

```python
# Questions

Check whether Bob's attendance is available.
Find students whose attendance has not been recorded.
Count present students.
Count absent students.
```

```python
#Solution:

if attendance.get("Bob") is not None:
    print("Bob's attendance is available")
else:
    print("Not available")

no_attendance = [name for name, attendances in attendance.items() if attendances is None]
print("Students with missing attendance: ", no_attendance)

present_count = 0
absent_count = 0
for name, attendances in attendance.items():
    if attendances is not None:
        if attendances:
            present_count = present_count + 1
        else:
            absent_count = absent_count + 1

print("No. of present Students: ",present_count)
print("No. of absent Students: ",absent_count)
```

```python
#Solution Q25, Q26, Q27, Q28, Q29:

#Q25.Increase the salary of every employee by 10%.
#Q26.Update the department of Employee 1040 to Research.
#Q27.Add a new employee to the dictionary.
#Q28.Delete Employee 1018 from the dictionary.
#Q29.Update the phone number of Employee 1075.

for empid, details in employees.items():
    details["Salary"] = details["Salary"]*1.10

employees["1040"]["Department"] = "Research"
print(employees["1040"]["Department"])

employees["1101"] = {"Name": "Employee_1101",
        "Department": "Operations",
        "Age": 26,
        "Salary": 68383,
        "Email": "employee1101@company.com",
        "Phone": "9812244967",
        "Manager": "Manager_13"}

del employees["1018"]

employees["1040"]["Phone"] = "9816739446"
```

```python
#Solution Q22, Q23, Q24:

#Q22.Count how many employees belong to each department.
#Q23.Count how many employees are managed by each manager.
#Q24.Find which department has the largest number of employees.

deptDict = {}
for empid, details in employees.items():
    if details.get("Department") in deptDict:
        deptDict[details.get("Department")] = deptDict.get(details.get("Department")) + 1
    else:
        deptDict[details.get("Department")] = 1

print("Department Count: ", deptDict)

managerDict = {}
for empid, details in employees.items():
    if details.get("Manager") in managerDict:
        managerDict[details.get("Manager")] = managerDict.get(details.get("Manager")) + 1
    else:
        managerDict[details.get("Manager")] = 1

print("\nManager Count: ", managerDict)

 
countOfEmps = 0
for deptName, empCount in deptDict.items():
    if empCount > countOfEmps:
        deptWithMoreEmployees = deptName
        countOfEmps = empCount

print(f"\nDepartment with highest number of employees: {deptWithMoreEmployees}")
```

```python
#Solution Q8, Q9, Q10, Q11:

#Q8.Find all employees whose email address is missing (None).
#Q9.Find all employees whose phone number is missing.
#Q10.Count how many employees have incomplete records (missing email or phone).
#Q11.Replace all missing email addresses with: Not Available

missing_email = []
missing_phone = []
for empid, details in employees.items():
    if details.get('Email') is None:
        missing_email.append(details.get('Name'))
    if details.get('Phone') is None:
        missing_phone.append(details.get('Name'))

print("Missing email: ", missing_email)
print("Missing phone: ", missing_phone)

missing_items_set = missing_email + missing_phone
print(f"{len(missing_items_set)} employees have incomplete records")

for empid, details in employees.items():
    if details.get('Email') is None:
        details['Email'] = "Not Available"

print(employees.get('1007'))
```

```python
#Solution Q12, Q13, Q14, Q15, Q16:

#Q12.Find all employees working in the IT department.
#Q13.Find all employees working in the Finance department.
#Q14.Display the names of all employees managed by Manager_4.
#Q15.Find all employees older than 50 years.
#Q16.Find all employees younger than 30 years.

itDeptEmployees = []
financeDeptEmployees = []
for empid, details in employees.items():
    if details.get("Department") == "IT":
        itDeptEmployees.append(details.get("Name"))
    elif details.get("Department") == "Finance":
        financeDeptEmployees.append(details.get("Name"))

print("Employees of IT Department: ", itDeptEmployees)
print("\nEmployees of Finance Department: ", financeDeptEmployees)

manager4Employees = []
for empid, details in employees.items():
    if details.get("Manager") == "Manager_4":
        manager4Employees.append(details.get("Name"))

print("\nEmployees working under Manager_4: ", manager4Employees)

olderThan50 = []
youngerThan30 = []

for empid, details in employees.items():
    if details.get("Age") > 50:
        olderThan50.append(details.get("Name"))
    elif details.get("Age") < 30:
        youngerThan30.append(details.get("Name"))

print("\nEmployees older than 50 years: ", olderThan50)
print("\nEmployees younger than 30 years: ", youngerThan30)
```

```python
#Solution Q19, Q20, Q21:

#Q19.Calculate the average salary of all employees.
#Q20.Display employees earning more than ₹100,000.
#Q21.Display employees earning less than ₹50,000.
sum = 0
for empid, details in employees.items():
    sum = sum + details.get("Salary")

avg_sal = sum / len(employees)

moreThan100000 = []
lessThan50000 = []
for empid, details in employees.items():
    if details.get("Salary") > 100000:
        moreThan100000.append(details.get("Name"))
    elif details.get("Salary") < 50000:
        lessThan50000.append(details.get("Name"))

print(f"Average Salary: {avg_sal} \n\nEmployees earning more than 100000: {moreThan100000} \n\nEmployees earning less than 50000: {lessThan50000}")
   
```
