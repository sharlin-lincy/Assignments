```python
# Question 9  

Create a dictionary of five products.

Example

    Laptop
    Mouse
    Keyboard
    Monitor
    Printer

Each product should contain

    Product Name
    Price
    Quantity

Write a program to

    Display all products.
    Print the product with the highest price.
    Calculate the total inventory value

(Formula:  Inventory Value = Price × Quantity)
```

```python
products = {"Laptop":{"ProductName":"Dell", "Price":23000, "Quantity":5}, "Mouse":{"ProductName":"Lenovo", "Price":249, "Quantity":3}, "Keyboard": {"ProductName":"Zebronics", "Price":452, "Quantity":8}, "Monitor":{"ProductName":"Apple", "Price":9990, "Quantity":4}, "Printer":{"ProductName":"hp", "Price":34570, "Quantity":7}}

print(products)

#Highest Price
highestPrice = 0
for product,details in products.items():
    for key,value in details.items():
        if key == "Price" and value > highestPrice:
            highestPrice = value

for product,details in products.items():
    for key,value in details.items():
        if key == "Price" and value == highestPrice:
            print(f"\nProduct with highest price is: {product}")

#Inventory Value
sum = 0
for product,details in products.items():
    sum = sum + details["Price"] * details["Quantity"]
print(f"\nTotal inventory value is: {sum}")
            
```

```python
# Question 10

Create a dictionary of five students and their marks.

Example

    Alice : 85
    Bob : 92
    Charlie : 78
    David : 88
    Emma : 95

Write a Python program to

    Find the student with the highest marks.
    Find the student with the lowest marks.
    Calculate the average marks.
    Display students who scored more than 85.
```

```python
students = {"Alice": 85, "Bob": 92, "Charlie": 78, "David": 88, "Emma": 95}

#HighestMarks
highestMark = 0
for mark in students.values():
    if mark > highestMark:
        highestMark = mark

print(highestMark)
for key,val in students.items():
    if val == highestMark:
        print(f"{key} has the highest mark")

#LowestMark
lowestMark = 100
for mark in students.values():
    if mark < lowestMark:
        lowestMark = mark

for key,val in students.items():
    if val == lowestMark:
        print(f"{key} has the lowest mark")

#AverageMarks
sum = 0
for mark in students.values():
    sum = sum + mark
avg = sum / len(students)
print(f"The average is: {avg}")

#More than 85
for name,mark in students.items():
    if mark > 85:
        print(f"{name} has scored more than 85")
```
